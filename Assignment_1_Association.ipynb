```python
import pandas as pd

df = pd.read_csv("Student_Performance_Dataset.csv")

print(df.head())
print(df.shape)
print(df.columns)
```

```python

# Assignment 1: Correlation Analysis

import matplotlib.pyplot as plt

# Select independent variables and dependent variable
columns = [
    "Study_Hours",
    "Attendance_Percent",
    "Previous_Marks",
    "Sleep_Hours",
    "Assignment_Score",
    "Final_Marks"
]

# Calculate correlation matrix
correlation = df[columns].corr()

print("Correlation Matrix:")
print(correlation.round(2))

# Display relationship with Final Marks
print("\nCorrelation with Final Marks:")
print(correlation["Final_Marks"].drop(
    "Final_Marks"
).sort_values(ascending=False).round(2))

# Draw heatmap
plt.figure(figsize=(9, 6))

plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")

plt.xticks(range(len(columns)), columns, rotation=45, ha="right")
plt.yticks(range(len(columns)), columns)

for i in range(len(columns)):
    for j in range(len(columns)):
        plt.text(
            j, i, f"{correlation.iloc[i, j]:.2f}",
            ha="center", va="center"
        )

plt.title("Correlation Heatmap of Student Performance")
plt.tight_layout()
plt.show()
```

## Assignment 1: Association between Variables

**Aim:** To identify the association between independent and dependent variables using correlation analysis.

**Result:** The relationship between independent variables and Final Marks was analysed using correlation analysis. A correlation matrix and heatmap were generated.
