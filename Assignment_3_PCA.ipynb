```python

# Assignment 3: Principal Component Analysis (PCA)

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 1: Load the same dataset
df = pd.read_csv("Student_Performance_Dataset.csv")

# Step 2: Select numerical features
features = [
    "Study_Hours",
    "Attendance_Percent",
    "Previous_Marks",
    "Sleep_Hours",
    "Assignment_Score",
    "Final_Marks"
]

X = df[features]

# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Apply PCA to reduce 6 features to 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Step 5: Create a new DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=["Principal_Component_1", "Principal_Component_2"]
)

print("Original dataset shape:", X.shape)
print("Reduced dataset shape:", pca_df.shape)

print("\nFirst 5 rows of reduced data:")
print(pca_df.head().round(3))

# Step 6: Display explained variance
print("\nExplained variance ratio:")
print(pca.explained_variance_ratio_.round(3))

print(
    "\nTotal variance retained:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)

# Step 7: Visualize the principal components
plt.figure(figsize=(8, 5))
plt.scatter(
    pca_df["Principal_Component_1"],
    pca_df["Principal_Component_2"]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Student Performance Dataset")
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Result:**

Principal Component Analysis (PCA) was successfully applied to the Student Performance Dataset. The six numerical features were transformed into two principal components, and the reduced data was visualized using a scatter plot.

**Conclusion:**

PCA reduced the dimensionality of the dataset from six numerical features to two components. This technique simplifies data representation while retaining a portion of the original data variation.
