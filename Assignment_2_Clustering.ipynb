```python

# Assignment 2: K-Means Clustering

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Step 1: Load the same saved dataset
df = pd.read_csv("Student_Performance_Dataset.csv")

# Step 2: Select features for clustering
features = [
    "Study_Hours",
    "Attendance_Percent",
    "Previous_Marks",
    "Final_Marks"
]

X = df[features]

# Step 3: Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Apply K-Means with 3 clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(X_scaled)

# Step 5: Display cluster assignments
print("First 10 students with cluster labels:")
print(df[features + ["Cluster"]].head(10))

print("\nNumber of students in each cluster:")
print(df["Cluster"].value_counts().sort_index())

# Step 6: Visualize the clusters
plt.figure(figsize=(8, 5))

for cluster in sorted(df["Cluster"].unique()):
    cluster_data = df[df["Cluster"] == cluster]

    plt.scatter(
        cluster_data["Study_Hours"],
        cluster_data["Final_Marks"],
        label=f"Cluster {cluster}"
    )

plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("K-Means Clustering of Students")
plt.legend()
plt.grid(True)
plt.show()
```

**Result:**

The K-Means clustering algorithm was successfully applied to the Student Performance Dataset. The students were divided into three clusters using study hours, attendance, previous marks and final marks. The clusters were visualized using a scatter plot.
